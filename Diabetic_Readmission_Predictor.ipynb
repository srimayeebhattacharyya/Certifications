# **About the Dataset**

The Diabetes 130-US Hospitals Dataset contains over 100,000 hospital admissions of diabetic patients collected from 130 hospitals in the United States between 1999 and 2008.

Each record represents one hospital encounter and contains demographic information, admission details, diagnoses, laboratory procedures, medications, length of stay, and the patient's readmission status.

The primary objective of this project is to clean and prepare the dataset for exploratory analysis and predictive modeling to identify the factors associated with hospital readmission.

## Cleaning Strategy Applied

Based on the initial data quality assessment and domain understanding, the following data cleaning and preparation steps were applied:

*   **Missing Value Identification**: Identified '?' characters as missing values and replaced them with `np.nan` for consistent handling across the dataset.
*   **Feature Enrichment**: Merged `diabetic_df` with the `IDS_mapping.csv` file to obtain descriptive names for `admission_type_id`, creating a new `admission_type_description` column. This enhances readability and interpretability.
*   **High Missing Value Columns**: Dropped columns with a very high percentage of missing values (`weight`, `max_glu_serum`, `A1Cresult`) as they contained insufficient information to be useful for analysis or imputation.
*   **Categorical Missing Value Imputation**: For remaining categorical columns with missing values (`race`, `payer_code`, `medical_specialty`, `diag_1`, `diag_2`, `diag_3`, `admission_type_description`), these `np.nan` values were imputed with 'Unknown' to preserve the records and signify their unknown status. This prevents loss of potentially valuable data rows.
*   **Inconsistent Data Handling**: Addressed the 'Not Mapped' value in the `admission_type_description` column by treating it similarly to other missing values, replacing it with 'Unknown'.

## Analytical Goals

This project aims to achieve the following analytical objectives based on the cleaned and preprocessed dataset:

*   **Identify Key Factors** — Determine which patient demographics, admission details, medical procedures, and medication usages are most significantly associated with hospital readmission for diabetic patients.
*   **Predictive Modeling** — Develop a robust predictive model to accurately forecast the likelihood of a patient being readmitted. The model will distinguish between readmission within 30 days, readmission after 30 days, and no readmission.
*   **Insights for Intervention** — Provide actionable insights that can inform healthcare providers and policy makers on potential interventions to reduce readmission rates and improve patient outcomes.
*   **Feature Importance** — Understand the relative importance of different features in predicting readmission, highlighting areas for targeted clinical focus.
